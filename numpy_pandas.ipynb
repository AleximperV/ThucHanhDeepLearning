```python
# Thêm dòng dữ liệu với hstack và vstack
# Hàm hstack((x, y)) cho phép thêm nối thêm dữ liệu ma trận y vào sau ma trận x (chiều ngang).
# Hàm vstack((x, y)) cho phép thêm nối thêm dữ liệu ma trận y vào dưới ma trận x (chiều dọc).
f = np.array([1.,2.,3.])
g = np.array([4.,5.,6.])
h_stack = np.hstack((f,g))
print('Nối ma trận theo chiều ngang\n',h_stack)
v_stack = np.vstack((f,g))
print('Nối ma trận theo chiều dọc\n', v_stack)
```

```python
# Bài 1: Tạo mảng 20 số nguyên ngẫu nhiên kích thước 3 x 3 có giá trị trong khoảng [0, 20]. Viết
# chương trình kiểm tra tất cả các phần tử trong mảng khác 0 (tìm hiểu hàm all của numpy)
mang_ngau_nhien = np.random.randint(0, 21, size=(3, 3))
print('Mảng ngẫu nhiên: \n',mang_ngau_nhien)
kiem_tra_so_0 = np.all(mang_ngau_nhien != 0)
print('Các phần tử trong mảng khác 0: ',kiem_tra_so_0)
```

```python
# Bài 2: Tạo mảng 20 số nguyên ngẫu nhiên kích thước 3 x 3 có giá trị trong khoảng [0, 20]. Viết
# chương trình kiểm tra tồn tại một phần tử trong mảng khác 0 (tìm hiểu hàm any của numpy)
mang_ngau_nhien = np.random.randint(0, 21, size=(3, 3))
print('Mảng ngẫu nhiên: \n',mang_ngau_nhien)
kiem_tra_so_0 = np.any(mang_ngau_nhien != 0)
print('Tồn tại một phần tử khác 0: ',kiem_tra_so_0)
```

```python
# Bài 3: Sinh viên tìm hiểu và viết chương trình sử dụng các hàm sau numpy.greater,
# numpy.greater_equal, numpy.less, và numpy.less_equal để so sánh 2 mảng 1 chiều, 2 chiều.
arr1_1 = np.random.randint(100, size=(8,))
arr1_2 = np.random.randint(100, size=(8,))

print('So sánh ngẫu nhiên: \n')
print('Mảng ngẫu nhiên 1: ', arr1_1)
print('Mảng ngẫu nhiên 2: ', arr1_2)

greater_compare_1 = np.greater(arr1_1,arr1_2)
print('So sánh: arr1_1 > arr1_2: ', greater_compare_1)

greater_equal_compare_1 = np.greater_equal(arr1_1,arr1_2)
print('So sánh: arr1_1 >= arr1_2: ', greater_equal_compare_1)

less_compare_1 = np.less(arr1_1,arr1_2)
print('So sánh: arr1_1 < arr1: ', less_compare_1)

less_equal_compare_1 = np.less_equal(arr1_1,arr1_2)
print('So sánh: arr1_1 <= arr1_2: ', less_equal_compare_1)
```

```python
# Bài 4: Viết chương trình tạo mảng gồm 10 số 1, 10 số 0 và 10 số 5
mang_1 = np.ones(10)
mang_2 = np.zeros(10)
mang_3 = np.full(10,5)
mang = np.concatenate((mang_1,mang_2,mang_3))
print(mang)
```

```python
# Bài 5: Viết chương trình tạo mảng số nguyên chẳn từ [30, 70]
mang_so_nguyen_chan = np.arange(30,71,2)
# Chạy từ số 30 và kết thúc ở 71 (Tối đa 70) với bước nhảy là 2
print(mang_so_nguyen_chan)
```

```python
# Bài 6: Viết chương trình tạo ma trận đơn vị 3 x 3 (đường chéo chính bằng 1, các phần tử còn lại 0)
# (Tìm hiểu hàm identity của numpy)
# np.identity sẽ luôn tạo ma trận vuông, với đường chéo chính (Trái Cao xuống Phải Dưới) luôn là 1
# và các phần tử còn lại là 0
ma_tran_don_vi = np.identity(3)
print(ma_tran_don_vi)
```

```python
# Bài 7: Viết chương trình tạo mảng 10 phần tử trong khoảng [15, 55]. In các giá trị ngoài trừ phần tử
# đầu và cuối
mang_ngau_nhien_bai_7 = np.random.randint(15,56, size=(10,))
print(mang_ngau_nhien_bai_7)
# Cắt bắt đầu phần tử thứ 2 (index 0 đến index 1) và ngừng ở phần tử cuối cùng là -1
# nghĩa là phần tử -1 loại ra, kết thúc ở phần tử kế cuối
mang_cat= mang_ngau_nhien_bai_7[1:-1]
print(mang_cat)
```
